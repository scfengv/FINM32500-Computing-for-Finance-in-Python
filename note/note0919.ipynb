## Stack vs. Queue

**Stack (LIFO: Last In, First Out)**
- list
- Undo buffers
- Rollback mechanisms
- Nested function calls
- Operations: (all O(1))
  - `Push`: Add an item to the top
  - `Pop`: Remove the top item
  - `Peek`: View the top item without removing it

**Queue (FIFO: First In, First Out)**
- deque
- Order processing
- Event handling
- Task scheduling
- Operations: (all O(1))
  - `Enqueue`: Add an item to the end
  - `Dequeue`: Remove the front item
  - `Front`: View the front item without removing it

```python
# Stack
stack = []
stack.append("Trade 1")   # Push
stack.append("Trade 2")
last_trade = stack.pop()  # Pop
print(stack[-1])   # Peek  # Trade 1
print(last_trade)  # LIFO  # Trade 2

# Queue
from collections import deque
queue = deque()
queue.append("Order 1")  # Enqueue
queue.append("Order 2")
first_order = queue.popleft()  # Dequeue
print(queue[0])     # Peek  # Order 2
print(first_order)  # FIFO  # Order 1
```

## Hash Maps & Sets

**Hash Maps**
- dict
- Key-value pairs
  - Order ID -> Order details
  - Asset -> Position size
  - Time stamp -> Portfolio snapshot
- Fast lookups, insertions, deletions (average O(1))
- Operations: (average O(1))
  - `Insert`: Add a key-value pair
  - `Delete`: Remove a key-value pair
  - `Lookup`: Retrieve a value by key

**Sets**
- set
- Unique items
  - Tracking active assets
  - Deduplicating trade IDs
  - Filtering unique signals
- Fast membership testing (average O(1))
- Operations: (average O(1))
  - `Add`: Insert an item
  - `Remove`: Delete an item
  - `Contains`: Check if an item exists

```python
# Hash Maps
trade_dict = {
    "T1": {"symbol": "AAPL", "quantity": 50},
    "T2": {"symbol": "GOOGL", "quantity": 30}
}
print(trade_dict["T1"]["symbol"])  # AAPL

# Sets
active_traders = {"Alice", "Bob", "Charlie"}
active_traders.add("Diana")
active_traders.remove("Bob")
print("Charlie" in active_traders)  # True
```

## Heaps & Priority Queues

**Heaps**
- heapq
- Binary tree structure where highest (or lowest) priority element is at the root (top)
- Min-heap or max-heap structure
  - Min-heap: smallest element at the root
  - Max-heap: largest element at the root (with a minus sign)
- Operations: 
  - `Insert`: Add an item while maintaining heap property (O(log n))
  - `Delete Min/Max`: Remove the root item (O(log n))
  - `Peek`: View the root item without removing it (O(1))

**Priority Queues**
- A queue where each element has a priority
- Elements are dequeued based on priority rather than arrival order
    - Matching best bid/ask in order book
    - Processing high-priority trades first
    - Managing risk alerts
- Operations:
  - `Enqueue`: Add an item with a priority (O(log n))
  - `Dequeue`: Remove the highest-priority item (O(log n))
  - `Peek`: View the highest-priority item without removing it (O(1))

```python
# Min Heaps
import heapq

heap = []
heapq.heappush(heap, 5)
heapq.heappush(heap, 2)
heapq.heappush(heap, 8)
heapq.heappush(heap, 1)

print(heap)         # Underlying tree structure (not sorted)
print(heap[0])      # Peek → 1
print(heapq.heappop(heap))  # Remove min → 1
print(heapq.heappop(heap))  # Remove next min → 2

# Max Heaps
import heapq

heap = []
heapq.heappush(heap, -5)
heapq.heappush(heap, -2)
heapq.heappush(heap, -8)
heapq.heappush(heap, -1)

print(heap)         # Underlying tree structure (not sorted)
print(-heap[0])      # Peek → 8
print(-heapq.heappop(heap))  # Remove max → 8
print(-heapq.heappop(heap))  # Remove next max → 5
```

```python
# Priority Queues
import heapq

ask_heap = []  # (price, order_id)

heapq.heappush(ask_heap, (100.5, "ask1"))
heapq.heappush(ask_heap, (100.2, "ask2"))
heapq.heappush(ask_heap, (101.0, "ask3"))

# Best ask (lowest price)
best_ask = ask_heap[0]
print("Best ask:", best_ask)

# Match order
matched = heapq.heappop(ask_heap)
print("Matched:", matched)
print("Next best ask:", ask_heap[0])
```

## Tree Structures

**Tree**
- A hierarchical data structure with nodes connected by edges
- Each node has: 
  - a value
  - a parent node
  - zero or more child nodes
- Common types:
  - Binary Trees (max 2 children per node)
  - Binary Search Trees (left child < parent < right child)
  - Balanced Trees (e.g., AVL, Red-Black Trees)
- Use cases:
  - Storing hierarchical data (e.g., organizational structures)
  - Efficient searching and sorting (e.g., binary search trees)
  - Implementing priority queues (e.g., heaps)
- Operations:
  - `Insert`: Add a node while maintaining tree properties (O(log n) for balanced trees)
  - `Delete`: Remove a node while maintaining tree properties (O(log n) for balanced trees)
  - `Search`: Find a node by value (O(log n) for balanced trees, BFS/DFS for general trees)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
```

## Graphs
- A collection of nodes (vertices) connected by edges
    - Directed vs. Undirected
    - Weighted vs. Unweighted
- Use cases:
  - Modeling networks (e.g., social networks, transportation networks)
  - Analyzing relationships (e.g., dependencies between assets)
  - Implementing algorithms (e.g., shortest path, network flow)
- Algorithms:
  - `Depth-First Search (DFS)`: **Explore** as far as possible along each branch before backtracking
  - `Breadth-First Search (BFS)`: **Explore** all neighbors at the present depth before moving on to nodes at the next depth level
  - `Dijkstra's Algorithm`: Find the **shortest path** from a source node to all other nodes in a weighted graph with non-negative weights
  - `A* Search`: Find the shortest path using heuristics to improve efficiency

| Data Structure         | Access              | Insert        | Delete        | Space        | Financial Use Case                               |

| **Stack** *(list)*     | O(1)                | O(1)          | O(1)          | O(n)         | Rollback logic, undo buffers                    |

| **Tree** *(BST)*       | O(log n)\*          | O(log n)      | O(log n)\*    | O(n)         | Sorted order books, asset taxonomy              |
| **Graph** *(DiGraph)*  | O(1) edge lookup    | O(1)          | O(1)          | O(n + e)     | Supply chain, strategy DAGs, market networks    |
| **Trie**               | O(k)                | O(k)          | O(k)          | O(k · n)     | Symbol lookup, prefix filtering                 |
| **Bloom Filter**       | O(k)                | O(k)          | ❌ (no delete) | O(m)         | High-speed deduplication, log filtering         |
