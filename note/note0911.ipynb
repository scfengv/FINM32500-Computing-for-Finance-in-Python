# Class vs Instance Variables in Python



```python
class Trader:
    firm = "ABC Capital"  # Class variable
    def __init__(self, name): # Constructor, set up the initial state
        self.name = name  # Instance variable
```

## Class Variable
- Defined outside any methods, usually at the top of the class.
- Shared across all instances of the class.
- Useful for constants, counters, or default settings.

## Instance Variable
- Defined inside methods using `self`.
- Unique to each instance of the class.
- Used to store data that varies between instances.
