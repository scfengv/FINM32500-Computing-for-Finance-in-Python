# Inheritance
- One class (child) inherits attributes and methods from another class (parent).
- Promotes code reuse and establishes a natural hierarchy.

````python
class Trader:
    def __init__(self, name, capital):
        self.name = name
        self.capital = capital

    def greet(self):
        print(f"Hello, I'm {self.name}")


class AlgoTrader(Trader):
    def __init__(self, name, capital, strategy):
        super().__init__(name, capital)
        self.strategy = strategy

    def run_strategy(self):
        print(f"Running {self.strategy} strategy")
````

## Method Overriding
- Subclass provides its own implementation of a method defined in the parent class.
- The subclass method replaces the parent class method when called on an instance of the subclass.
- Customization and Specialization

````python
class Trader:
    def greet(self):
        print("Hello, I'm a trader.")


class AlgoTrader(Trader):
    def greet(self):  # Overrides parent method
        print("Hello, I'm an algorithmic trader.")


t1 = Trader()
t2 = AlgoTrader()

t1.greet()  # Output: Hello, I'm a trader.
t2.greet()  # Output: Hello, I'm an algorithmic trader.
````

## Polymorphism
- Same method name, different behavior depending on the object.
- Treat different objects uniformly while allowing for specific implementations.

````python
class Trader:
    def execute(self):
        print("Executing manual trade")


class AlgoTrader(Trader):
    def execute(self):
        print("Executing algorithmic trade")


class HighFrequencyTrader(Trader):
    def execute(self):
        print("Executing high-frequency trade")


traders = [Trader(), AlgoTrader(), HighFrequencyTrader()]

for t in traders:
    t.execute()
````

## Abstract Base Classes
- A class that cannot be instantiated on its own.
- Serves as a **blueprint** for other classes.
- Defines methods that **must be created within any child classes** derived from the abstract base class.

````python
from abc import ABC, abstractmethod

class Trader(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def execute_trade(self, symbol, quantity):
        pass


class AlgoTrader(Trader):
    def execute_trade(self, symbol, quantity):  # Any subclass must implement `execute_trade`
        print(f"{self.name} executes algorithmic trade: {symbol}, {quantity}")

try:
    trader = Trader()  # This will raise an error
except TypeError as e:
    print(e)

algo_trader = AlgoTrader("AlgoBot")
algo_trader.execute_trade("AAPL", 100)
````

## Multiple Inheritance
- A class can inherit from more than one parent class.
- The subclass gains attributes and methods from all parent classes.
- Resolves conflicts using the Method Resolution Order (MRO).
  - The MRO determines the order in which base classes are searched when executing a method.

````python
class Logger:
    def log(self, msg):
        print(f"[LOG]: {msg}")


class Trader:
    def execute_trade(self):
        print("Trade executed")


class AlgoTrader(Trader, Logger):
        pass


at = AlgoTrader()
at.execute_trade()          # From Trader
at.log("Trade complete")    # From Logger
````

**MRO**

````python
class A:
    def greet(self):
        print("A")

class B:
    def greet(self):
        print("B")

class C(A, B):
    pass

c = C()
c.greet()
# Output: A
# C → A → B → object

class D(B, A):
    pass

d = D()
d.greet()
# Output: B
# D → B → A → object
````

````python
class A:
    def greet(self): print("A")

class B(A):
    def greet(self): print("B")

class C(A):
    def greet(self): print("C")

class D(B, C):
    pass

d = D()
d.greet()
# Output: B
# D → B → C → A → object
````

## Operator Overloading
- Customize the behavior of operators (like +, -, *, ==, etc.) for user-defined classes.
- Define special methods (dunder methods, like `__add__`, `__eq__`, etc.) in your class to specify how operators should behave.

| Operator | Method    | Description         |
|----------|-----------|----------------------|
| +        | `__add__` | Addition             |
| -        | `__sub__` | Subtraction          |
| ==       | `__eq__`  | Equality check       |
| <        | `__lt__`  | Less than            |
| >        | `__gt__`  | Greater than         |
| str()    | `__str__` | String representation |

````python
class Order:
    def __init__(self, symbol, quantity):
        self.symbol = symbol
        self.quantity = quantity

    # + Addition: combine quantities if same symbol
    def __add__(self, other):
        if self.symbol == other.symbol:
            return Order(self.symbol, self.quantity + other.quantity)
        raise ValueError("Cannot add orders with different symbols")

    # - Subtraction: reduce quantities if same symbol
    def __sub__(self, other):
        if self.symbol == other.symbol:
            return Order(self.symbol, self.quantity - other.quantity)
        raise ValueError("Cannot subtract orders with different symbols")

    # == Equality: same symbol & same quantity
    def __eq__(self, other):
        return (self.symbol == other.symbol) and (self.quantity == other.quantity)

    # < Less than: compare quantities (only allowed for same symbol)
    def __lt__(self, other):
        if self.symbol != other.symbol:
            raise ValueError("Cannot compare orders with different symbols")
        return self.quantity < other.quantity

    # > Greater than: compare quantities
    def __gt__(self, other):
        if self.symbol != other.symbol:
            raise ValueError("Cannot compare orders with different symbols")
        return self.quantity > other.quantity

    # str(): pretty printing
    def __str__(self):
        return f"{self.quantity} shares of {self.symbol}"


o1 = Order("AAPL", 100)
o2 = Order("AAPL", 50)
o3 = Order("AAPL", 100)

print("Addition:", o1 + o2)       # 150 shares of AAPL
print("Subtraction:", o1 - o2)    # 50 shares of AAPL

print("Equality:", o1 == o3)      # True
print("Less than:", o2 < o1)      # True
print("Greater than:", o1 > o2)   # True

print("String:", str(o1))         # "100 shares of AAPL"
````

## Python `@property` Decorator
- Used to define getter, setter, and deleter methods for class attributes.
- Turns a method into an attribute-like access.
- Protect internal data with validation logic.

### Getter and Setter
- treat attributes like variables, while still running code behind the scenes

**Getter**
- code that runs when you read an attribute

```Python
# Without Getter
print(obj.x) `  # Direct access to attribute x
```
```Python
# With Getter
@property
def x(self):
    return self._x  # Access internal attribute _x

print(obj.x)   # Calls a method internally that computes, validates, or transforms the value.
```

**Setter**
- code that runs when you assign a value to an attribute
  - Set some validation logic when setting a value
    - Validate the value
    - Transform the value
      - Log changes
    - Trigger other update rules
    - Maintain invariants

```Python
# Without Setter
obj.x = 10 # Directly assigns value
```

```Python
# With Setter
obj.x = 10 # Calls Setter method internally
```

````python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    # Getter method, read-only attribute
    # Run when accessing `fahrenheit` # t.fahrenheit
    @property
    def fahrenheit(self):
        return (self._celsius * 9/5) + 32

    # Setter method
    # Run when assigning to `fahrenheit` # t.fahrenheit = value
    @fahrenheit.setter
    def fahrenheit(self, value):
        self._celsius = (value - 32) * 5/9


t = Temperature(0)
print(t.fahrenheit)  # Get Fahrenheit  # Calls Getter -> returns 32.0

t.fahrenheit = 212   # Set Fahrenheit to 212  # Calls Setter -> Update _celsius to 100.0
print(t._celsius)    # 100.0
````

````python
class TradingAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance  # private attribute

    @property
    def balance(self):
        """Getter: returns the current balance."""
        print("Fetching balance...")
        return self._balance

    @balance.setter
    def balance(self, amount):
        """Setter: enforces rules before changing balance."""
        if amount < 0:
            raise ValueError("Balance cannot be negative!")
        print(f"Updating balance to {amount}...")
        self._balance = amount
        
account = TradingAccount("Alice", 1000)
print(account.balance)  # Call Getter  # Fetching balance... 1000
account.balance = 1500  # Call Setter  # Updating balance to 1500...
print(account.balance)  # Call Getter  # Fetching balance... 1500
````

# Exception Handling
- Mechanism to handle runtime errors gracefully.
- Use `try`, `except`, `else`, and `finally` blocks to manage exceptions

```Python
try:
    # risky code
    pass

except SpecificException as e:
    # error handling
    pass

else:
    # runs if no exception
    pass

finally:
    # always runs
    pass
```

**Exceptions**

| Exception           | When It Occurs                   |
|---------------------|----------------------------------|
| ZeroDivisionError   | Division by zero                 |
| FileNotFoundError   | Opening a non-existent file      |
| TypeError           | Invalid operation on a type      |
| ValueError          | Wrong value supplied             |

**Custom Exceptions**
- Subclass `Exception` to create your own exception types.
- Use `raise` to signal specific error conditions.
```Python
class TradeError(Exception):
    pass

def execute_trade(qty):
    if qty <= 0:
        raise TradeError("Quantity must be positive")
```
